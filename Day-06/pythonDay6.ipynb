## How to add elements in tuple

```python
# Approach 1
t1 = (10,20,30)
t2 = (40,50,60)
print(t1+t2) # concatenation of tuples
```

```python
t3 = (70,)
print(t1+t3) # concatenation of tuples
```

```python
# Approach 2

t1 = (10,20,30)
print(t1)
print(type(t1))
t1 = list(t1)
print(t1) 
print(type(t1))
t1.append(40) # adding new element to list
t1 = tuple(t1) # converting list back to tuple
print(t1)
```

```python
# Approach 3
t1 =(10,20,30,40,[50,60,70,80])
print(t1)
t1[4].append(90) # adding new element to list inside tuple
print(t1)
```

## Packing and unpacking

```python
# packing 
t1 = 10,20,30,40
print(t1)
```

```python
# Unpacking
t1,t2,t3 = (10,20,30)
print(t1)
print(t2)
print(t3)
```

## Set Declaration

```python
s1 ={10,20,30,"Innomatics",10,11,13}
print(s1)
```

```python
# creating empty set
s2 =set()
print(s2)
print(type(s2))
```

```python
# creating empty dictionary
d1 = {}
print(d1)
print(type(d1))
```

```python
# creating set using list 
l1 = [10,20,30,10,20,30]
s3 = set(l1)
print(s3)   
```

```python
s1[0]     # indexing is not possible in set
```

## Built in functions
 - .add()

```python
s3.add(100) # adding new element to set
print(s3)
```

```python
s3.add(50,60,70)
print(s3) # adding multiple elements to set is not possible
```

 - .update()

 adding multiple elements to set is possible using update method inside(list) pass

```python
s3.update([50,60,70]) 
print(s3)
```

 - .sort( )

```python
sorted(s3) 
```

```python
sorted(s3, reverse=True) 
```

 - .discard( )

 discard will discard the value fron the set
 
 if the value is not in a set then discard will not throw any error

```python
s3.discard(45) 
print(s3)
```

```python
s3.discard(10)
print(s3)
```

 - .remove ( )

 remove will remove the value nfrom the set if the value exists if not it generate error

```python
s3.remove(45)
```

```python
s3.remove(20)
print(s3)
```

 - .pop( )

 pop is used to delete element from the given set 

 we can not the predict the which element will be delete because set is unindexed

```python
s3.pop()
print(s3)
```

 - .clear( )

 remove all elements in a set empty set returns

```python
s3.clear()
print(s3)
```
