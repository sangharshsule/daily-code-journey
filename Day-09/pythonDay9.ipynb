## Control Flow Statements
- pass:used in place of a block which is required syntactically but not logically

```python
# pass

for i in range(10):
               if i==5:
                              pass
               print(i)
               
```

- break : terminate the loop together

```python
# break

for i in range(10):
               if i==5:
                              break
               print(i)
```

- continue :except that value without loop breakes next values return

```python
# continue

for i in range(10):
               if i==5:
                              continue
               print(i)
```

```python
# while pass
i = 0
while i<10:
               if i==5:
                              pass
               print(i)
               i+=1
```

```python
# while break
i = 0
while i<10:
               if i==5:
                              break
               print(i)
               i+=1
```

```python
# while continue
#Approach 1
i = 0

while i < 10:
    if i == 5:
        i += 1
        continue

    print(i)
    i += 1
```

```python
# Approach 2
i =0 
while i<10:
               i+=1
               if i == 5:
                              continue
               print(i)
```
