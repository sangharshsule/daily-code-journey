##### Q. Create a Python program to store the details of 5 employees (Employee ID, Employee Name, Department, Age, and Salary) in a dictionary. Take Employee ID or Employee Name as input from the user and display the corresponding employee details. If the employee is not found, print "Enter Valid Input".
    

```python
# get the employee details based on the index position and fetch the details of all employees 

user_input = input("Enter Employee Name / ID / Department: ")



if user_input in emp["emp_name"]:

    index_pos = emp["emp_name"].index(user_input)



elif  int(user_input) in emp["emp_id"]:

    index_pos = emp["emp_id"].index(int(user_input))



elif user_input in emp["dept"]:

    index_pos = emp["dept"].index(user_input)

else:

    print("Employee Not Found")

    exit()

print("Employee ID:", emp["emp_id"][index_pos])

print("Employee Name:", emp["emp_name"][index_pos])

print("Department:", emp["dept"][index_pos])

print("Age:", emp["age"][index_pos])

print("Salary:", emp["emp_sal"][index_pos])
    
```

## 5) copy()

- original copy also change by using shallow

```python
# to create one more instance of a list

import copy

l=[[1,2,3,4],[5,6,7,8]]

shallow =copy.copy(l)

print(l)
print(shallow)
```

```python
shallow[0][0]=21

print(shallow)
print(l)
```

## 6) deepcopy ()

- original copy not change 

```python
og=[[1,2,3,4],[5,6,7,8]]

deep =copy.deepcopy(og)

print(og)
print(deep)
```

```python
deep[0][0]=21

print(deep)
print(og)
```
