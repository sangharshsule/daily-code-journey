# Operators in Expressions

# Arithmatic Operators

```python
x=15
y=4
print('x + y =',x+y) #add
print('x - y =',x-y) #sub
print('x * y =',x*y) #mul
print('x / y =',x/y) #div
print('x // y =',x//y) #floor
print('x ** y =',x**y) #power
print('x % y =',x%y) #modulus
print(y**.2)
```

## Logical Operators

```python
x=True
y=False
print('x and y is',x and y)
print('x or y is',x or y)
print('not x is',not x)
```

## Identity Operators

```python
# it checks the in memory location of the variable
x1 = 5
y1 =5
x2 ='Hello'
y2 ='Hello'
x3 =[1,2,3]
y3 =[1,2,3]
print(x1 is y1) #true
print(x2 is y2) #true
print(x3 is y3) #false
```

## Assignment Operators

```python
x=15
y=4
print("X =",x)
#x=x+y
x+=y
print('x += y =',x)
x-=y
print('x -= y =',x)
x*=y
print('x *= y =',x)
x/=y
print('x /= y =',x)
x//=y
print('x //= y =',x)
x**=y
print('x **= y =',x)
```

## Membership Operators

###### (in and not in) are the membership operators in python they are used to test whether a value or variable is found in a sequence (string,list,tuple,set,Dictionary)

```python

x ='Hello World'
y ={1:'a',2:'b',3:'c'}

print('h' in x)
print('hello' not in x)
print(1 in y)
```

## Comparison Operators

```python
x=15
y=15

print("x==y =>",x==y)
print("x!=y =>",x!=y)
print("x>y =>",x>y)
print("x<y =>",x<y)
print("x>=y =>",x>=y)
print("x<=y =>",x<=y)
```

## Python Conditional Statements

###### Conditions help us to make decisions they execute blocks of code based on truthfulliness of some statements.The if ,if-else,elif ,nested if and if-else,elif ladder keywords actually help us to write conditions

```python
pos=-12
if pos>=0:
    print("Positive number")
```

```python
if pos>=0:
    print("Positive number")
else:
    print("Negative number")
```

```python
pos=12
if pos>=0:
    print("Positive number")
```

```python
age = 209
if age >= 18 and age <= 100:
    print("The price is 1400")
elif age > 9 and age < 18:
    print("The price is 900")
elif age > 3 and age <= 9:
    print("The price is 500")
elif age>0 and age <= 3:
    print("The Childrens below 3 years are not allowed to enter the park")
else:
    print("Invalid age, please enter a valid age")
```
