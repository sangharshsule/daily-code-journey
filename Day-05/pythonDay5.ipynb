## 1.append()
append add single value at the end of list
 
Use case: Dynamically build a list from user input or results.

```python
l=[1,2,3,4,5,6,7]
print(l)
```

```python
l.append(8)
print(l)
```

## 2.extend()
extend () is used to extend our current list with another list

```python
l.extend([9,10,11])
print(l)
```

```python
l.extend([12,13])
print(l)
```

## pop()

Use case: When you want to remove and use an item simultaneously.

```python
# to remove the last element from the list we can use pop() method
l.pop()
```

## 3.remove()
remove is used to remove a value from the list if the value is known but not its index

Use case: Cleaning up or filtering specific elements.

```python
l.remove(6)
print(l)
```

Note: the pop() method is used to remove values from a list when their index positions are known and remove() is used when their face value is known
