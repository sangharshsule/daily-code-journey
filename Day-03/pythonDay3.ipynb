## Nested if-else

```python
age=int(input("Enter your age: "))
if age>18:
    has_license=input("Do you have a driving license? (yes/no): ")
    if has_license.lower()=="yes":
        print("You are eligible to drive.")
    else:
        print("You need a driving license to drive.")
else:
    print("You are not eligible to drive.")
```

```python
has_license=bool(int(input("Do you have a driving license?: "))) #logic to convert 1/0 to True/False
print("You have a driving license:", has_license)
```

## Ladder elif

```python
marks=float(input("Enter your marks: "))

if marks>90 and marks<=100:
    print("You got an A+ grade.")
elif marks>80 and marks<=90:
    print("You got an A grade.")
elif marks>70 and marks<=80:
    print("You got a B+ grade.")
elif marks>60 and marks<=70:
    print("You got a B grade.")
elif marks>50 and marks<=60:
    print("You got a C+ grade.")
elif marks>=35 and marks<=50:
    print("You got a C grade.")
else:
    print("You failed the exam.")
```

## Bitwise Operator

```python
x = 6 #0110
y = 3 #0011
print('x & y:', x & y)  # Bitwise AND
print('x | y:', x | y)  # Bitwise OR
print('x ^ y:', x ^ y)  # Bitwise XOR       
print('~x:', ~x)      # Bitwise NOT
print('x << 2:', x << 2)  # Left Shift
print('x >> 2:', x >> 2)  # Right Shift
```
