## Understanding range() in python

- Syntax : range(start,stop,step)
- the range fun generate a sequence of numbers.
- used commonly in loops
- if one argument is given its treated as the stop.

```python
# range(start,stop(excluded),step)
list(range(0,10,1))
```

- useful for controlled iteration,repeating tasks, or indexing

```python
for i in range(0,11,1):
               print(i)
```

```python
for i in range(0,11,1):
               if i%2==0:
                              print("Even nu is :",i)
               else:
                              print("Odd nu is :",i)
        
```
