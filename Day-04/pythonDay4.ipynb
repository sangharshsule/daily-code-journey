## Problem Statement
##### A customer support bot receives messy inputs.You want to clean up the text it contains keywords like "refund".
##### Steps to solve
##### 1.Take a sample user Message.
##### 2.Lowercase it and strip extra whitespaces.
##### 3.Check if it contains the word "refund".

```python
s=input("Enter Message :")
print(s.lower())
print(s.strip())
print(s.split("."))
print(s.lower().count("refund"))

if "refund" in s.lower():
    print("Refund is available")
else:
    print("Refund is not available")
```
